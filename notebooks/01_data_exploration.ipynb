```python
# Import Required Libraries

# Used for handling and analysing tabular data
import pandas as pd

# Used for numerical operations and arrays
import numpy as np

# Used for creating graphs and visualisations
import matplotlib.pyplot as plt

# Provides cleaner and more advanced statistical visualisations
import seaborn as sns

# Helps manage file paths in a cleaner and more portable way
from pathlib import Path



# Display & Visual Settings


# Display all dataframe columns when viewing the dataset
pd.set_option("display.max_columns", None)

# Apply a cleaner visual style for plots and graphs
sns.set(style="whitegrid")
```

```python
# Load the dataset from the raw data folder
DATA_PATH = Path("../data/raw/PCOS_data.csv")

# Read the CSV file into a pandas dataframe
df = pd.read_csv(DATA_PATH)

# Display the first 5 rows of the dataset
df.head()
```

```python
# Display the number of rows and columns in the dataset
df.shape
```

```python
# Display dataset structure and data types
df.info()
```

```python
# Generate descriptive statistics for numerical features
df.describe()
```

```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Sort missing values from highest to lowest
missing_values = df.isnull().sum().sort_values(ascending=False)

# Display only columns with missing values
missing_values[missing_values > 0]
```

```python
# Display class distribution for PCOS diagnosis
df["PCOS (Y/N)"].value_counts()
```

```python
# Create a visualisation of PCOS class distribution
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="PCOS (Y/N)")

# Add graph title and labels
plt.title("Distribution of PCOS Diagnosis")
plt.xlabel("PCOS Diagnosis")
plt.ylabel("Number of Patients")

# Display the graph
plt.show()
```

```python
# Generate a correlation matrix for numerical features
correlation_matrix = df.corr(numeric_only=True)

# Display correlation heatmap
plt.figure(figsize=(18,12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    annot=False
)

# Add title
plt.title("Correlation Heatmap of Numerical Features")

# Display heatmap
plt.show()
```

```python
# Calculate correlations with the PCOS target variable
pcos_correlations = correlation_matrix["PCOS (Y/N)"].sort_values(ascending=False)

# Display correlations with PCOS
pcos_correlations
```

```python
# Select the top correlations with PCOS (excluding the target itself)
top_correlations = pcos_correlations[1:11]

# Create bar chart of top correlated features
plt.figure(figsize=(10,6))

top_correlations.plot(kind="bar")

# Add title and labels
plt.title("Top Features Correlated with PCOS")
plt.xlabel("Features")
plt.ylabel("Correlation Coefficient")

# Rotate feature labels for readability
plt.xticks(rotation=45)

# Display the graph
plt.show()
```
